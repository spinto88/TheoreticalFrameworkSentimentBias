### Analizar datos 

A partir de las distribuciones de menciones positivas, neutral y negativas se calculan los parametros del modelo de ideología. 

```python
# Esta direccion es la IP de setubal en la red del DF
# El 8000 es un puerto de setubal 
# el endpoint marca qué se hace (/analyze en este caso)
url = "http://10.99.32.142:8000/analyze"

headers = {
    "Content-Type": "application/json"
}

# Llamado
response_analyze = requests.post(url, json=data, headers=headers)

# Respuesta
response_analyze_json = response_analyze.json()
```
